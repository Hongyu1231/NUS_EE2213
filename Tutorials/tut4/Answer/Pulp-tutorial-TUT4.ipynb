# Example: Solving a Linear Programming Problem with PuLP

A small furniture workshop produces wooden chairs (x) and wooden tables (y).
They aim to maximize their total profit for the week, where:

Each chair sold earns $3 in profit.

Each table sold earns $2 in profit.

Resource Constraints (This Week):

Wood Supply:
Each chair requires 2 units of wood.
Each table requires 1 unit of wood.
The total wood available is 100 units.

Labor Hour:
Each chair requres 1 hour of labour, and each table also requires 1 hour of labor.
The total available labor time is 80 hours for the week.

### Example: Solving a Linear Programming Problem with PuLP
We will solve the following problem using PuLP:

**Objective:** Maximize profit: Z = 3x + 2y

**Subject to constraints:**
- 2x + y ≤ 100  (resource constraint)
- x + y ≤ 80    (time constraint)
- x ≥ 0, y ≥ 0  (non-negativiy)


````python
# Step 1: Import PuLP
from pulp import *
````

````python
# Step 2: Define the LP problem
prob = LpProblem(name='maximize_profit', sense=LpMaximize)
````

````python
# Step 3: Define the decision variables
x = LpVariable(name='x', lowBound=0)
y = LpVariable(name='y', lowBound=0)
````

````python
# Step 4: Add the objective function
prob += 3 * x + 2 * y
````

````python
# Step 5: Add constraints
prob += 2 * x + y <= 100  # Wood resource constraint
prob +=  x +  y <= 80   # Labour hour constraint'
````

````python
# Step 6: Solve the model
prob.solve()
````

````python
# Results
x_opt = value(x) # Get the optimal value of decision variable x
y_opt = value(y) # Get the optimal value of decision variable y
max_profit = value(prob.objective) # Get the maximum profit from the objective function
````

````python
# Print results
print("Optimal number of units to produce:")
print("Product chairs:", x_opt)
print("Product tables:", y_opt)
print("Maximum Profit: $", max_profit)
````
